```python
"""
Diabetes Regression Analysis Hw
This Assignment loads the scikit learn diabetes dataset, trains three distinct 
regression models, and evaluates their performance to determine the best fit.
"""

import sys
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python

class DiabetesRegressionAnalysis:
    
    # Class used to process loading data, training models,and evaluating machine learning regressions on the Diabetes dataset.


    def __init__(self, test_size=0.2, random_state=42):
        "defines the three models."
        self.test_size = test_size
        self.random_state = random_state
        self.X_train, self.X_test, self.y_train, self.y_test = None, None, None, None
        
        # Dictionary of models to train. 
        self.models = {
            "Ridge Regression": Ridge(alpha=1.0, random_state=self.random_state),
            "Random Forest": RandomForestRegressor(n_estimators=100, random_state=self.random_state),
            "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, random_state=self.random_state)
        }
        self.results = {}

    def prepare_data(self):
        "Loads the diabetes dataset, scales features, and splits into train/test sets."
        try:
            print("preparing the diabetes dataset")
            # Load the dataset
            diabetes = load_diabetes()
            X = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
            y = diabetes.target

            # Split the data
            X_train_raw, X_test_raw, self.y_train, self.y_test = train_test_split(
                X, y, test_size=self.test_size, random_state=self.random_state
            )

            #Standardize features to improve model performance and stability
            scaler = StandardScaler()
            self.X_train = scaler.fit_transform(X_train_raw)
            self.X_test = scaler.transform(X_test_raw)
            print("Data successfully loaded and prepared.")

        except Exception as e:
            print(f"Err0r: Failed to prepare data.")
            sys.exit(1)

    def train_and_evaluate(self):
        # Trains all models in the dictionary and evaluates it
        if self.X_train is None:
            print("Err0r: Data has not been prepared yet.")
            return

        print("Training and evaluating models")
        for name, model in self.models.items():
            try:
                # Train the model
                model.fit(self.X_train, self.y_train)
                
                # Make predictions
                predictions = model.predict(self.X_test)
                
                # Calculate metrics
                mse = mean_squared_error(self.y_test, predictions)
                rmse = np.sqrt(mse)
                mae = mean_absolute_error(self.y_test, predictions)
                r2 = r2_score(self.y_test, predictions)
                
                # Store results
                self.results[name] = {
                    "MSE": mse,
                    "RMSE": rmse,
                    "MAE": mae,
                    "R2": r2
                }
                print(f"[{name}] successfully trained.")
            except Exception as e:
                print(f"Err0r: Failed to train or evaluate {name}.")

    def display_results(self):
        #Prints the evaluation metrics for all trained models and identifies the best one
        if not self.results:
            print("No results to display. Run train_and_evaluate() first.")
            return

        print("Model Evaluation Results")

        best_model_name = None
        best_r2 = -float('inf')

        for name, metrics in self.results.items():
            print(f"\n{name}:")
            print(f"Mean Squared Error (MSE): {metrics['MSE']:.2f}")
            print(f"Root Mean Squared Error (RMSE): {metrics['RMSE']:.2f}")
            print(f"Mean Absolute Error (MAE): {metrics['MAE']:.2f}")
            print(f"R-squared (R2): {metrics['R2']:.4f}")

            # Get the best model based on R squared
            if metrics['R2'] > best_r2:
                best_r2 = metrics['R2']
                best_model_name = name

        print(f"Best Performing Model: {best_model_name}")

def main():
    print("Starting Diabetes Regression Analysis.")
    
    analysis = DiabetesRegressionAnalysis()
    
    
    analysis.prepare_data()
    analysis.train_and_evaluate()
    analysis.display_results()

if __name__ == "__main__":
    main()
```
